# Eksperimen Sederhana: Seed, `rng.random()`, dan `rng.integers()`

Tujuannya hanya melihat nilai yang dihasilkan generator pseudoacak NumPy.


```python
import numpy as np
import pandas as pd

SEED = 42
print(f'Seed: {SEED}')
```

## 1. Dua generator dengan seed sama

Keduanya dimulai dari seed yang sama, sehingga urutan nilainya sama.


```python
rng_1 = np.random.default_rng(SEED)
rng_2 = np.random.default_rng(SEED)

nilai_1 = rng_1.random(8)
nilai_2 = rng_2.random(8)

hasil_random_seed_sama = pd.DataFrame({
    'pemanggilan_ke': range(1, 9),
    'rng_1': nilai_1,
    'rng_2': nilai_2,
    'sama': np.isclose(nilai_1, nilai_2),
})
display(hasil_random_seed_sama)
```

Nilai dari pemanggilan 1 sampai 8 berbeda. Namun, nilai pada baris yang sama dari `rng_1` dan `rng_2` identik karena seed-nya sama.


## 2. Nilai berubah pada setiap pemanggilan

Satu generator yang sama tidak mengeluarkan angka identik terus-menerus. Ia bergerak ke nilai berikutnya dalam urutan pseudoacak.


```python
rng = np.random.default_rng(SEED)

for pemanggilan_ke in range(1, 9):
    print(f'pemanggilan {pemanggilan_ke}: {rng.random():.8f}')
```

## 3. `rng.integers()` dengan seed sama

Contoh ini menghasilkan bilangan bulat dari 0 sampai 2. Pada proyek rute lari, bentuk seperti ini dipakai untuk memilih indeks action legal ketika agen melakukan eksplorasi.


```python
rng_1 = np.random.default_rng(SEED)
rng_2 = np.random.default_rng(SEED)

integer_1 = rng_1.integers(0, 3, size=12)
integer_2 = rng_2.integers(0, 3, size=12)

hasil_integers_seed_sama = pd.DataFrame({
    'pemanggilan_ke': range(1, 13),
    'rng_1_integers': integer_1,
    'rng_2_integers': integer_2,
    'sama': integer_1 == integer_2,
})
display(hasil_integers_seed_sama)
```

## 4. Generator tanpa seed

Dua generator tanpa seed eksplisit biasanya menghasilkan urutan berbeda.


```python
rng_tanpa_seed_1 = np.random.default_rng()
rng_tanpa_seed_2 = np.random.default_rng()

nilai_tanpa_seed_1 = rng_tanpa_seed_1.integers(0, 3, size=12)
nilai_tanpa_seed_2 = rng_tanpa_seed_2.integers(0, 3, size=12)

hasil_tanpa_seed = pd.DataFrame({
    'pemanggilan_ke': range(1, 13),
    'generator_1': nilai_tanpa_seed_1,
    'generator_2': nilai_tanpa_seed_2,
    'sama': nilai_tanpa_seed_1 == nilai_tanpa_seed_2,
})
display(hasil_tanpa_seed)
```

## Kesimpulan

Dengan seed sama, urutannya dapat diulang. Namun setiap pemanggilan tetap menghasilkan nilai berikutnya yang berbeda. Tanpa seed, urutannya biasanya berubah setiap generator dibuat.
